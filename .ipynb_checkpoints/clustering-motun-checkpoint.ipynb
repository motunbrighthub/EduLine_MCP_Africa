```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\USER\EduLine_MCP_Africa\mcqss.csv",
    sep=",",
    engine="python",
    on_bad_lines="skip"   # skips problematic rows
)
print(df.head())
print(df.columns)
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
embeddings = model.encode(df["Question"].tolist(), show_progress_bar=True)
```

```python
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(embeddings)
```

```python
from sklearn.cluster import KMeans

k = 9  
kmeans = KMeans(n_clusters=k, random_state=42)
labels = kmeans.fit_predict(embeddings)
```

```python
df["Cluster"] = labels
df["PCA1"] = X_pca[:, 0]
df["PCA2"] = X_pca[:, 1]


plt.figure(figsize=(10, 6))
sns.scatterplot(x="PCA1", y="PCA2", hue="Cluster", data=df, palette="tab10", s=50)
plt.title("Sentence Transformer + PCA Clustering of WAEC Questions")
plt.show()
```

```python
import random
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# ==========================================
# 🧠 Function: Adaptive MCQ System
# ==========================================
def adaptive_mcq_system(df, num_clusters=9, batch_size=5, performance_history=None):
    """
    Adaptive MCQ practice engine.
    
    Parameters:
    - df: DataFrame with columns ['Question', 'Option A', 'Option B', 'Option C', 'Option D', 'Answer']
    - num_clusters: Number of clusters to form (default = 9)
    - batch_size: Number of questions per round (default = 5)
    - performance_history: dict to track accuracy per cluster {cluster_id: score}
    
    Returns:
    - next_batch: DataFrame of next questions to attempt
    - performance_history: Updated dictionary of user performance
    """

    # Step 1: Generate sentence embeddings (if not already clustered)
    if "Cluster" not in df.columns:
        print("Clustering questions...")
        model = SentenceTransformer("all-MiniLM-L6-v2")
        embeddings = model.encode(df["Question"].astype(str).tolist(), show_progress_bar=True)
        kmeans = KMeans(n_clusters=num_clusters, random_state=42)
        df["Cluster"] = kmeans.fit_predict(embeddings)
        
        # For visualization (optional)
        pca = PCA(n_components=2, random_state=42)
        df["PCA1"], df["PCA2"] = pca.fit_transform(embeddings).T
    
    # Step 2: Initialize performance history if not provided
    if performance_history is None:
        performance_history = {i: 0.0 for i in range(num_clusters)}
    
    # Step 3: Identify weakest clusters (lowest accuracy)
    weakest_clusters = sorted(performance_history.items(), key=lambda x: x[1])[:3]
    weak_ids = [cid for cid, _ in weakest_clusters]
    
    # Step 4: Choose next batch — 70% from weak clusters, 30% random
    weak_batch = df[df["Cluster"].isin(weak_ids)].sample(int(batch_size * 0.7), replace=True)
    random_batch = df.sample(int(batch_size * 0.3), replace=True)
    
    next_batch = pd.concat([weak_batch, random_batch]).sample(frac=1).reset_index(drop=True)
    
    print("\n✅ Generated next batch of questions:")
    for i, row in next_batch.iterrows():
        print(f"\nQ{i+1}: {row['Question']}")
        print(f"A. {row['Option A']}\nB. {row['Option B']}\nC. {row['Option C']}\nD. {row['Option D']}")
    
    return next_batch, performance_history
```

```python
from sklearn.cluster import KMeans
from sentence_transformers import SentenceTransformer
import pandas as pd


waec_mcqs = pd.read_csv(
    r"C:\Users\USER\EduLine_MCP_Africa\mcqss.csv",
    sep=",",
    engine="python",
    on_bad_lines="skip"   # skips problematic rows
)
print(waec_mcqs.head())
print(waec_mcqs.columns)

```

```python
waec_mcqs = waec_mcqs.dropna(subset=["Question"])
waec_mcqs = waec_mcqs[waec_mcqs["Question"].str.strip() != ""]
```

```python
# Load Sentence Transformer
model = SentenceTransformer("all-MiniLM-L6-v2")

# Store results
all_results = []
valid_subjects = ["Mathematics", "English", "Chemistry", "Physics"]

df = df[df["Subject"].isin(valid_subjects)]
```

```python
# Loop through each subject
for subject, group in waec_mcqs.groupby("Subject"):
    print(f"\n📘 Clustering subject: {subject} ({len(group)} questions)")
    
    # Generate sentence embeddings for that subject
    embeddings = model.encode(group["Question"].tolist(), show_progress_bar=True)
    
    # Decide number of clusters (you can tweak this per subject)
    n_clusters = min(9, len(group))  # or pick based on content
    
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    labels = kmeans.fit_predict(embeddings)
    
    # Save cluster labels
    group["Cluster"] = labels
    all_results.append(group)
```

```python
valid_subjects = ["Mathematics", "English", "Chemistry", "Physics"]

# Keep only rows where Subject matches exactly
waec_mcqs = waec_mcqs[waec_mcqs["Subject"].isin(valid_subjects)]

print("✅ Cleaned data shape:", waec_mcqs.shape)
print("🧠 Subjects left:", waec_mcqs["Subject"].unique())
```

```python
# Combine all clustered subjects
waec_mcqs_clustered = pd.concat(all_results)

# Save the subject-based clusters
waec_mcqs_clustered.to_csv("subject_based_clusters.csv", index=False)

print("\n✅ Clustering complete! File saved as subject_based_clustersing.csv")
```
